Общие сведения об инструкциях

```python
import os
import fitz # PyMuPDF

pdf_directory = "Instructions"

with_text_layer = 0
without_text_layer = 0
total_files = 0

for pdf_file in os.listdir(pdf_directory)
    if pdf_file.endswith(".pdf"):
        total_files += 1
        pdf_path = os.path.join(pdf_directory, pdf_file)
        
        doc = fitz.open(pdf_path)
        
        has_text = False
        for page in doc:
            text = page.get_text().strip()
            if text:  # если текст не пустой
                has_text = True
                break
        
        if has_text:
            with_text_layer += 1
        else:
            without_text_layer += 1

print(f"Всего файлов: {total_files}")
print(f"С текстовым слоем: {with_text_layer}")
print(f"Без текстового слоя: {without_text_layer}")
```

Получение названий файлов (инструкций), у которых есть текстовый слой и у которых его нет

```python
import os
import fitz  # PyMuPDF

pdf_directory = "Instructions"

files_with_text_layer = []
files_without_text_layer = []

for pdf_file in os.listdir(pdf_directory):
    if pdf_file.endswith(".pdf"):
        pdf_path = os.path.join(pdf_directory, pdf_file)
        
        doc = fitz.open(pdf_path)
        has_text_layer = any(page.get_text().strip() for page in doc)
        
        if has_text_layer:
            files_with_text_layer.append(pdf_file)
        else:
            files_without_text_layer.append(pdf_file)

with open("with_text_layer.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(files_with_text_layer))

with open("without_text_layer.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(files_without_text_layer))

print(f"Файлы с текстовым слоем: {len(files_with_text_layer)}")
print(f"Файлы без текстового слоя: {len(files_without_text_layer)}")
print("Списки сохранены в 'with_text_layer.txt' и 'without_text_layer.txt'.")
```

Получение инструкций, у которых есть текстовый слой, с помощью PyMuPDF.

```python
import os
import fitz  # PyMuPDF
import json

pdf_directory = "Instructions"  
output_directory = "parsed_instructions"  
os.makedirs(output_directory, exist_ok=True)

def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text_sections = []
    for page_num, page in enumerate(doc, start=1):
        text = page.get_text().strip()
        if text:
            text_sections.append({"page": page_num, "content": text})
    return text_sections

if __name__ == "__main__":
    processed_files = 0

    for pdf_file in os.listdir(pdf_directory):
        if pdf_file.endswith(".pdf"):
            pdf_path = os.path.join(pdf_directory, pdf_file)
            
            doc = fitz.open(pdf_path)
            has_text_layer = any(page.get_text().strip() for page in doc)
            if not has_text_layer:
                continue

            processed_files += 1

            # JSON-структура
            parsed_data = {
                "file_name": pdf_file,
                "text_sections": extract_text_from_pdf(pdf_path)
            }

            output_path = os.path.join(output_directory, pdf_file.replace(".pdf", ".json"))
            with open(output_path, "w", encoding="utf-8") as json_file:
                json.dump(parsed_data, json_file, ensure_ascii=False, indent=4)

            print(f"[{processed_files}] Обработан файл: {pdf_file}. JSON сохранён в: {output_path}")

    print(f"Всего обработано файлов с текстовым слоем: {processed_files}")
```

Получение инструкций, у которых нет текстового слоя, с помощью TesseractOCR.

```python
import os
import fitz  # PyMuPDF
import pytesseract
from PIL import Image
import json

pdf_directory = "Instructions" 
output_directory = "ocr_instructions"  
os.makedirs(output_directory, exist_ok=True)

# Tesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

def extract_text_with_ocr(pdf_path):
    doc = fitz.open(pdf_path)
    ocr_results = []
    for page_num in range(len(doc)):
        pix = doc[page_num].get_pixmap(dpi=1100)  
        img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)

        text = pytesseract.image_to_string(img, lang="rus")
        ocr_results.append({"page": page_num + 1, "text": text})

    return ocr_results

if __name__ == "__main__":
    processed_files = 0

    for pdf_file in os.listdir(pdf_directory):
        if pdf_file.endswith(".pdf"):
            pdf_path = os.path.join(pdf_directory, pdf_file)

            doc = fitz.open(pdf_path)
            has_text_layer = any(page.get_text().strip() for page in doc)
            if has_text_layer:
                continue

            processed_files += 1

            ocr_results = extract_text_with_ocr(pdf_path)

            # JSON-структура
            parsed_data = {
                "file_name": pdf_file,
                "ocr_text_sections": ocr_results
            }

            output_path = os.path.join(output_directory, pdf_file.replace(".pdf", ".json"))
            with open(output_path, "w", encoding="utf-8") as json_file:
                json.dump(parsed_data, json_file, ensure_ascii=False, indent=4)

            print(f"[{processed_files}] Обработан файл: {pdf_file}. JSON сохранён в: {output_path}")

    print(f"Всего обработано файлов без текстового слоя: {processed_files}")
```

Метаданные для инструкций с текстовым слоем. Метаданные для файла 2743e0d0-5d1f-4d23-b0c4-893dc416c2ac.pdf (Аспирин Кардио) не найдены (он вообще отсутствует в файле с метаданными). 

```python
import os
import json
import pandas as pd

json_directory = "parsed_instructions" 
metadata_file = "instr_metadata_updated.tsv"  
output_directory = "updated_instructions"  
os.makedirs(output_directory, exist_ok=True)

metadata = pd.read_csv(metadata_file, sep="\t", header=None)
metadata.columns = ["mnn", "trade_name", "drug_id", "pdf_path"]  

for json_file in os.listdir(json_directory):
    if json_file.endswith(".json"):
        json_path = os.path.join(json_directory, json_file)

        with open(json_path, "r", encoding="utf-8") as f:
            parsed_data = json.load(f)

        file_metadata = metadata[metadata["pdf_path"].str.contains(parsed_data["file_name"], na=False)]
        if file_metadata.empty:
            print(f"Метаданные для файла {parsed_data['file_name']} не найдены. Пропущен.")
            continue

        updated_data = {
            "mnn": file_metadata["mnn"].values[0],
            "trade_name": file_metadata["trade_name"].values[0],
            "drug_id": file_metadata["drug_id"].values[0],
            "file_name": parsed_data["file_name"],
            "text_sections": parsed_data.get("text_sections", [])
        }

        output_path = os.path.join(output_directory, json_file)
        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(updated_data, f, ensure_ascii=False, indent=4)

        print(f"Метаданные для файла: {parsed_data['file_name']}")

print("Процесс завершён.")
```

Удаление дубликатов

```python
import os

instructions_dir = "updated_instructions"
ocr_instructions_dir = "updated_ocr_instructions"

instructions_files = set(os.listdir(instructions_dir))
ocr_instructions_files = set(os.listdir(ocr_instructions_dir))

common_files = instructions_files & ocr_instructions_files

for file_name in common_files:
    instructions_path = os.path.join(instructions_dir, file_name)
    ocr_instructions_path = os.path.join(ocr_instructions_dir, file_name)

    if os.path.exists(instructions_path):
        os.remove(instructions_path)
        print(f"Удалён дубликат из {instructions_dir}: {file_name}")

    if os.path.exists(ocr_instructions_path):
        os.remove(ocr_instructions_path)
        print(f"Удалён дубликат из {ocr_instructions_dir}: {file_name}")

print(f"Удалено {len(common_files)} дубликатов из обеих папок.")
```
