```python
import os
import json
import pandas as pd
from IPython.display import HTML, display
```

```python
folders = [
    "updated_instructions",
    "updated_ocr_instructions"
]

records = []

for folder in folders:
    for filename in os.listdir(folder):
        if not filename.lower().endswith(".json"):
            continue
        full_path = os.path.join(folder, filename)
        with open(full_path, "r", encoding="utf-8") as f:
            data = json.load(f)
        # удаляем символ '®'
        mnn = data.get("mnn", "").replace("®", "")
        trade_name = data.get("trade_name", "").replace("®", "")
        drug_id = data.get("drug_id", "").replace("®", "")
        file_name = data.get("file_name", "").replace("®", "")

        records.append({
            "mnn": mnn,
            "trade_name": trade_name,
            "drug_id": drug_id,
            "file_name": file_name
        })

df = pd.DataFrame(records, columns=["mnn", "trade_name", "drug_id", "file_name"])

df.sort_values(by="mnn", inplace=True)

df.reset_index(drop=True, inplace=True)

df.to_csv("instructions_metadata.csv", index=False, encoding="utf-8-sig")

print(f"Собрано записей: {len(df)}")
```

**Таблица с количеством инструкций по каждому MNN**

```python
pd.set_option('display.max_rows', None)

df = pd.read_csv("instructions_metadata.csv", encoding="utf-8-sig")

df["mnn"] = df["mnn"].astype(str).str.strip()
df["trade_name"] = df["trade_name"].astype(str).str.strip()

trade_counts = (
    df
    .groupby("mnn")["trade_name"]
    .count()
    .sort_values(ascending=False)
    .reset_index(name="trade_name_count")
)

avg_trade_per_mnn = trade_counts["trade_name_count"].mean()
print(f"Среднее число торговых названий на одно MNN: {avg_trade_per_mnn:.2f}\n")

display(trade_counts)
```
